```python
import cv2
import matplotlib.pyplot as plt
```

```python
# Step 1: Load the image
image = cv2.imread('EX4.jpg')  # Load the image from file
```

```python
# Display the original image
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for correct display
plt.title("Original Image")  
plt.axis('off') 
```

```python
import cv2
import numpy as np

# image should already be loaded
rows, cols = image.shape[:2]

# Shift the image by (100, 50)
M = np.float32([[1, 0, 100],
                [0, 1, 50]])

translated_image = cv2.warpAffine(image, M, (cols, rows))
import matplotlib.pyplot as plt

plt.imshow(cv2.cvtColor(translated_image, cv2.COLOR_BGR2RGB))
plt.title("Translated Image")
plt.axis("off")
plt.show()
```

```python
# Step 3: Image Scaling
fx, fy = 5.0, 2.0  # Scaling factors (1.5x scaling for both width and height)
scaled_image = cv2.resize(image, None, fx=fx, fy=fy, interpolation=cv2.INTER_LINEAR)
# resize: Resize the image by scaling factors fx, fy
# INTER_LINEAR: Uses bilinear interpolation for resizing
plt.imshow(cv2.cvtColor(scaled_image, cv2.COLOR_BGR2RGB))  # Display the scaled image
plt.title("Scaled Image")  # Set title
plt.axis('off')
```

```python
# Step 4: Image Shearing
import cv2
import numpy as np

rows, cols = image.shape[:2]

# Shearing matrix
M = np.float32([
    [1, 0.5, 0],
    [0, 1, 0]
])

sheared_image = cv2.warpAffine(image, M, (int(cols * 1.5), rows))
```

```python
import matplotlib.pyplot as plt

plt.imshow(cv2.cvtColor(sheared_image, cv2.COLOR_BGR2RGB))
plt.title("Sheared Image")
plt.axis("off")
plt.show()
```

```python
# Step 5: import cv2
import matplotlib.pyplot as plt

# Load the building image
image = cv2.imread("EX4.jpg")

# Reflect (flip horizontally)
reflected_image = cv2.flip(image, 1)

# Display
plt.imshow(cv2.cvtColor(reflected_image, cv2.COLOR_BGR2RGB))
plt.title("Reflected Building Image")
plt.axis("off")
plt.show()
```

```python
# Step 6: Image Rotation
(height, width) = image.shape[:2]  # Get the image height and width
angle = 45  # Rotation angle in degrees (rotate by 45 degrees)
center = (width // 2, height // 2)  # Set the center of rotation to the image center
M_rotation = cv2.getRotationMatrix2D(center, angle, 1)  # Get the rotation matrix
# getRotationMatrix2D: Takes the center of rotation, angle, and scale factor (1 means no scaling)
rotated_image = cv2.warpAffine(image, M_rotation, (width, height))  # Apply rotation
```

```python
plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))  # Display the rotated image
plt.title("Rotated Image")  # Set title
plt.axis('off')
```

```python
# Step 7: Image Cropping
x, y, w, h = 100, 100, 200, 150  # Define the top-left corner (x, y) and the width (w) and height (h) of the crop
# Cropping the image from coordinates (x, y) to (x+w, y+h)
cropped_image = image[y:y+h, x:x+w]
# The crop is performed by slicing the image array in the y and x directions
```

```python
plt.imshow(cv2.cvtColor(cropped_image, cv2.COLOR_BGR2RGB))  # Display the cropped image
plt.title("Cropped Image")  # Set title
plt.axis('off')
```
